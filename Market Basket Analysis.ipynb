```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Load your data
df = pd.read_csv('your_retail_data.csv')

# Create the basket (Only for one country to keep it fast, e.g., 'United Kingdom')
basket = (df[df['Country'] =="United Kingdom"]
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))

# Convert to 1s and 0s
def encode_units(x):
    if x <= 0: return 0
    return 1
basket_sets = basket.applymap(encode_units)

# Find patterns
frequent_itemsets = apriori(basket_sets, min_support=0.03, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)

# Result
rules.sort_values('lift', ascending=False).head(10)
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# 1. Load your Excel file
# Make sure the name matches 'sql project.xlsx' exactly as seen in your folder
df = pd.read_excel('sql project.xlsx')

# --- ADD THE FILTER HERE ---
df = df[df['Country'] == 'United Kingdom']
# ---------------------------

# 2. Basic Cleaning: Remove rows without an Invoice Number
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
df['InvoiceNo'] = df['InvoiceNo'].astype('str')

# 3. Create the "Basket"
# We group by Invoice and Description to see what items were bought together
basket = (df.groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))

# 4. Convert to 1s and 0s (True/False)
# Association rules only care IF an item was bought, not how many
def encode_units(x):
    if x <= 0: return 0
    return 1

basket_sets = basket.map(encode_units)

# Check the first few rows of your basket
basket_sets.head()
```

```python
# 1. Find items that appear together in at least 2% of transactions
# (If this results in an empty table, change 0.02 to 0.01)
frequent_itemsets = apriori(basket_sets, min_support=0.02, use_colnames=True)

# 2. Generate the association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)

# 3. Sort by the strongest "Lift" (The most meaningful correlations)
rules = rules.sort_values('lift', ascending=False)

# 4. View your results!
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10)
```
